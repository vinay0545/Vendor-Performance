```python
import pandas as pd
import os
from sqlalchemy import create_engine
import logging
import time

logging.basicConfig(
    filename="logs/ingestion_db.log",
    level=logging.DEBUG,
    format="%(asctime)s - %(levelname)s - %(message)s",
    filemode="a"
)

engine = create_engine('sqlite:///inventory.db')

def ingest_db(df, table_name, engine, if_exists='append'):
    df.to_sql(
        table_name,
          con=engine,
            if_exists=if_exists,
              index=False,
              chunksize=10000
              )
    
def load_raw_data():
    start = time.time()

    for file in os.listdir('data'):
        if file.endswith('.csv'):
            table_name = file[:-4] # Remove the .csv extension to get the table name
            first_chunk = True
            for df in pd.read_csv('data/' + file, chunksize=50000):
                print("Loading:",table_name, df.shape)
                if first_chunk:
                 ingest_db(df, table_name, engine, 'replace')
                 first_chunk = False
                else:
                    ingest_db(df, table_name, engine,'append')
    end = time.time()
    total_time = (end - start) / 60

    print("Ingestion Completed")
    print("Total Time Taken:",total_time, "minutes")
load_raw_data()
```
